```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt

# 1. Crear la señal de prueba con ruido
fs = 1000  # Frecuencia de muestreo
t = np.linspace(0, 1.0, fs, endpoint=False)

# Señal limpia (combinación de frecuencias)
senial_limpia = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 50 * t)
# Ruido blanco
np.random.seed(42)
ruido = np.random.normal(0, 0.5, t.shape)
senial_con_ruido = senial_limpia + ruido

# 2. Definición de Filtros Butterworth
def filtro_pasabajos(data, cutoff, fs, order=4):
    nyq = 0.5 * fs
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    return filtfilt(b, a, data)

def filtro_pasaaltos(data, cutoff, fs, order=4):
    nyq = 0.5 * fs
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='high', analog=False)
    return filtfilt(b, a, data)

def filtro_pasabandas(data, lowcut, highcut, fs, order=4):
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    b, a = butter(order, [low, high], btype='band', analog=False)
    return filtfilt(b, a, data)

# 3. Aplicar los filtros
filtrada_pasa_bajos = filtro_pasabajos(senial_con_ruido, cutoff=10, fs=fs)
filtrada_pasa_altos = filtro_pasaaltos(senial_con_ruido, cutoff=40, fs=fs)
filtrada_pasa_bandas = filtro_pasabandas(senial_con_ruido, lowcut=40, highcut=60, fs=fs)

# 4. Graficar resultados
plt.figure(figsize=(10, 8))

plt.subplot(4, 1, 1)
plt.plot(t, senial_con_ruido, color='gray', label='Señal con Ruido')
plt.plot(t, senial_limpia, color='black', label='Señal Limpia')
plt.title('1. Señal de Entrada vs Señal Limpia')
plt.legend(loc='upper right')
plt.grid(True)

plt.subplot(4, 1, 2)
plt.plot(t, filtrada_pasa_bajos, color='blue', label='Pasa Bajos (Fc = 10 Hz)')
plt.title('2. Filtro Pasa Bajos (Elimina alta frecuencia y ruido)')
plt.legend(loc='upper right')
plt.grid(True)

plt.subplot(4, 1, 3)
plt.plot(t, filtrada_pasa_altos, color='red', label='Pasa Altos (Fc = 40 Hz)')
plt.title('3. Filtro Pasa Altos (Elimina frecuencias bajas)')
plt.legend(loc='upper right')
plt.grid(True)

plt.subplot(4, 1, 4)
plt.plot(t, filtrada_pasa_bandas, color='green', label='Pasa Bandas (40-60 Hz)')
plt.title('4. Filtro Pasa Bandas (Aísla componente central)')
plt.legend(loc='upper right')
plt.grid(True)

plt.tight_layout()
plt.show()
```
